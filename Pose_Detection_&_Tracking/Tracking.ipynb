```python
from ultralytics import YOLO

model=YOLO('yolov8n.pt')

file_path = '/content/drive/MyDrive/Colab Notebooks/Computer Vision/Pose_detection/football.mp4'

results = model.predict(source=file_path, show=True, save=True, conf=0.4, project='/content/drive/MyDrive/Colab Notebooks/Computer Vision/Pose_detection')

```

```python
class_counts = {}

for box in results[0].boxes:

  cls = int(box.cls)

  class_name = model.names[int(cls)]

  class_counts[class_name] = class_counts.get(class_name, 0) + 1

for class_name, count in class_counts.items():
  print(f'{class_name}: {count}')
```
